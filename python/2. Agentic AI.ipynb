```python
import pandas as pd
import random
from datetime import datetime, timedelta
```

```python
random.seed(42)
```

```python
DATA_START = pd.Timestamp("2026-01-01")
DATA_END = pd.Timestamp("2026-09-30")
```

```python
CUSTOMER_START = pd.Timestamp("2026-01-01")
CUSTOMER_END = pd.Timestamp("2026-08-31")
```

Table Creation
1. Customers


```python
first_names = [
    "Aarav", "Aditi", "Aisha", "Akash", "Ananya","Arjun", "Diya", "Ishaan", "Kabir", "Kavya","Kiara", "Meera", "Neha", "Pooja", "Rahul",
    "Rhea", "Rohan", "Saanvi", "Simran", "Tanvi"
]

last_names = [
    "Sharma", "Verma", "Gupta", "Mehta", "Kapoor","Malhotra", "Jain", "Agarwal", "Khanna", "Bansal"
]
```

```python
customers = []

for i in range(1, 201):

    platform_user_id = f"PLAT{900000 + i}"

    name = f"{random.choice(first_names)} {random.choice(last_names)}"

    created_at = CUSTOMER_START + pd.Timedelta(
        days=random.randint(
            0,
            (CUSTOMER_END - CUSTOMER_START).days
        )
    )

    customers.append({
        "platform_user_id": platform_user_id,
        "name": name,
        "created_at": created_at
    })

customers_df = pd.DataFrame(customers)
```

```python
print('head:',customers_df.head())
print('shape:',customers_df.shape)
print('duplicates:',customers_df["platform_user_id"].duplicated().sum())
print('nulls:',customers_df.isnull().sum())
```

Table Creation
2. Products


```python
categories = [
    "Dresses","Tops","Co-ords","Kurtas","Bottoms","Jackets"
]

materials_by_category = {
    "Dresses": [
        "Cotton", "Linen", "Rayon", "Viscose","Polyester", "Silk Blend", "Chanderi", "Georgette"
    ],
    "Tops": [
        "Cotton", "Linen", "Rayon", "Viscose","Polyester", "Silk Blend", "Knit"
    ],
    "Co-ords": [
        "Cotton", "Linen", "Rayon", "Viscose","Polyester", "Silk Blend"
    ],
    "Kurtas": [
        "Cotton", "Rayon", "Viscose","Silk Blend", "Chanderi"
    ],
    "Bottoms": [
        "Cotton", "Linen", "Viscose","Denim", "Polyester"
    ],
    "Jackets": [
        "Denim", "Cotton", "Polyester", "Knit"
    ]
}
```

```python
product_names = {
    "Dresses": [
        "Floral Midi Dress","Wrap Dress","A-Line Dress","Pleated Maxi Dress","Ruched Party Dress"
    ],
    "Tops": [
        "Cotton Shirt","Ribbed Top","Satin Blouse","Printed Top","Crop Shirt"
    ],
    "Co-ords": [
        "Linen Co-ord Set","Printed Co-ord Set","Relaxed Co-ord Set","Embroidered Co-ord Set","Party Co-ord Set"
    ],
    "Kurtas": [
        "Straight Kurta","A-Line Kurta","Printed Kurta","Embroidered Kurta","Festive Kurta"
    ],
    "Bottoms": [
        "Wide-Leg Trousers","Straight Pants","Palazzo Pants","Denim Jeans","Pleated Skirt"
    ],
    "Jackets": [
        "Denim Jacket","Cropped Jacket","Utility Jacket","Quilted Jacket","Lightweight Blazer"
    ]
}

prices = [
    799, 999, 1199, 1299, 1499,1699, 1899, 1999, 2199,2499, 2799, 2999, 3299, 3499
]
```

```python
products = []

for i in range(1, 101):

    category = random.choice(categories)
    base_name = random.choice(product_names[category])

    products.append({
        "product_id": f"P{i:03d}",
        "product_name": f"{base_name} {i:02d}",
        "category": category,
        "price": random.choice(prices),
        "material": random.choice(materials_by_category[category]),
        "active_status": random.choices(
            ["Active", "Inactive"],
            weights=[85, 15]
        )[0]
    })

products_df = pd.DataFrame(products)
```

```python
print('head:',products_df.head(10))
print('shape:',products_df.shape)
print('category count:',products_df["category"].value_counts())
print('status count:',products_df["active_status"].value_counts())
print('product_id duplicates:',products_df["product_id"].duplicated().sum())
```

Table Creation
3. Product Variants


```python
colors = [
    "Black", "White", "Blue", "Green", "Pink","Red", "Beige", "Brown", "Yellow", "Grey"
]

sizes = ["XS", "S", "M", "L", "XL"]
```

```python
product_variants = []
variant_counter = 1

for _, row in products_df.iterrows():

    product_id = row["product_id"]

    selected_colors = random.sample(colors, 2)

    for colour in selected_colors:

        for size in sizes:

            stock_qty = random.randint(0, 20)

            product_variants.append({
                "variant_id": f"V{variant_counter:04d}",
                "product_id": product_id,
                "colour": colour,
                "size": size,
                "stock_qty": stock_qty
            })

            variant_counter += 1

variants_df = pd.DataFrame(product_variants)
```

```python
variants_df.head(10)
```

```python
variants_df.shape
```

```python
print("Duplicate variant IDs:",
      variants_df["variant_id"].duplicated().sum())

print("All product IDs valid:",
      variants_df["product_id"].isin(
          products_df["product_id"]
      ).all())

print("Out of stock:",
      (variants_df["stock_qty"] == 0).sum())

print("Low stock:",
      variants_df["stock_qty"].between(1, 5).sum())

print("Normal stock:",
      (variants_df["stock_qty"] > 5).sum())
```

Table Creation
4. Orders

```python
reference_date = pd.Timestamp("2026-09-30")

orders = []

for i in range(1, 501):

    customer = customers_df.sample(1).iloc[0]

    platform_user_id = customer["platform_user_id"]
    customer_created_at = customer["created_at"]

    # 15% recent active-order candidates
    is_recent = random.random() < 0.15

    if is_recent:
        order_start = max(
            customer_created_at,
            reference_date - pd.Timedelta(days=7)
        )

        order_date = order_start + pd.Timedelta(
            days=random.randint(
                0,
                (reference_date - order_start).days
            )
        )

        days_old = (reference_date - order_date).days

        if days_old <= 1:
            order_status = random.choice(["Placed", "Processing"])

        elif days_old <= 3:
            order_status = random.choice(["Processing", "Shipped"])

        elif days_old <= 5:
            order_status = random.choice(["Shipped", "Out for Delivery"])

        else:
            order_status = random.choice(
                ["Out for Delivery", "Delivered"]
            )

    else:
        order_date = customer_created_at + pd.Timedelta(
            days=random.randint(
                0,
                (reference_date - customer_created_at).days
            )
        )

        order_status = random.choices(
            ["Delivered", "Cancelled"],
            weights=[85, 15]
        )[0]

    expected_delivery_date = (
        order_date + pd.Timedelta(days=random.randint(3, 8))
    )

    orders.append({
        "order_id": f"ORD{i:04d}",
        "platform_user_id": platform_user_id,
        "order_date": order_date,
        "order_status": order_status,
        "expected_delivery_date": expected_delivery_date
    })

orders_df = pd.DataFrame(orders)
```

```python
orders_df.head(10)
```

```python
print("Shape:", orders_df.shape)

print(
    orders_df["order_status"].value_counts()
)

print(
    "Orders before customer creation:",
    (
        orders_df.merge(
            customers_df[
                ["platform_user_id", "created_at"]
            ],
            on="platform_user_id"
        )["order_date"]
        <
        orders_df.merge(
            customers_df[
                ["platform_user_id", "created_at"]
            ],
            on="platform_user_id"
        )["created_at"]
    ).sum()
)
```

```python
active_statuses = [
    "Placed",
    "Processing",
    "Shipped",
    "Out for Delivery"
]

old_active_orders = orders_df[
    orders_df["order_status"].isin(active_statuses)
    &
    (
        orders_df["order_date"]
        <
        reference_date - pd.Timedelta(days=7)
    )
]

print("Old active orders:", len(old_active_orders))
```

```python
print(
    "Duplicate order IDs:",
    orders_df["order_id"].duplicated().sum()
)

print(
    "Null values:"
)
print(orders_df.isnull().sum())
```

Table Creation
5. Order Items


```python
order_items = []
order_item_counter = 1

for order_id in orders_df["order_id"]:

    number_of_items = random.randint(1, 3)

    selected_variants = variants_df.sample(
        n=number_of_items,
        replace=False
    )

    for _, variant in selected_variants.iterrows():

        variant_id = variant["variant_id"]
        product_id = variant["product_id"]

        quantity = random.randint(1, 2)

        unit_price = products_df.loc[
            products_df["product_id"] == product_id,
            "price"
        ].iloc[0]

        order_items.append({
            "order_item_id": f"OI{order_item_counter:05d}",
            "order_id": order_id,
            "variant_id": variant_id,
            "quantity": quantity,
            "unit_price": unit_price
        })

        order_item_counter += 1

order_items_df = pd.DataFrame(order_items)
```

```python
order_items_df.head(10)
```

```python

print("Shape:", order_items_df.shape)

print(
    "Duplicate order_item_id:",
    order_items_df["order_item_id"].duplicated().sum()
)

print(
    "All order IDs valid:",
    order_items_df["order_id"].isin(
        orders_df["order_id"]
    ).all()
)

print(
    "All variant IDs valid:",
    order_items_df["variant_id"].isin(
        variants_df["variant_id"]
    ).all()
)

print(order_items_df.isnull().sum())
```

Table creation
6. Messages

```python
intents = [
    "Product Information","Inventory Check","Order Tracking","Shipping Query","Human Support"
]
```

```python
message_templates = {
    "Product Information": [
        "What is the price of this product?",
        "What material is this?",
        "What colours are available?",
        "Can you tell me more about this item?"
    ],

    "Inventory Check": [
        "Is this available in M?",
        "Do you have this in black?",
        "Is size L available?",
        "Is this currently in stock?"
    ],

    "Order Tracking": [
        "Where is my order?",
        "Can you check my order status?",
        "Has my order been shipped?",
        "When will my order arrive?"
    ],

    "Shipping Query": [
        "How long does delivery usually take?",
        "How many days does shipping take?",
        "Do you deliver across India?",
        "What is the usual delivery time?"
    ],

    "Human Support": [
        "I want to return my order.",
        "I want a refund.",
        "My product arrived damaged.",
        "I received the wrong item.",
        "I want to cancel my order.",
        "Please connect me with customer support."
    ]
}
```

```python
messages = []

for i in range(1, 601):

    customer = customers_df.sample(1).iloc[0]

    platform_user_id = customer["platform_user_id"]
    customer_created_at = customer["created_at"]

    intent = random.choice(intents)

    message_text = random.choice(
        message_templates[intent]
    )

    message_date = customer_created_at + pd.Timedelta(
        days=random.randint(
            0,
            (DATA_END - customer_created_at).days
        ),
        hours=random.randint(0, 23),
        minutes=random.randint(0, 59)
    )

    if intent in ["Product Information", "Inventory Check"]:
        current_product_id = random.choice(
            products_df["product_id"].tolist()
        )
    else:
        current_product_id = None

    messages.append({
        "message_id": f"MSG{i:05d}",
        "platform_user_id": platform_user_id,
        "sender_type": "Customer",
        "message_text": message_text,
        "intent": intent,
        "timestamp": message_date,
        "current_product_id": current_product_id
    })

messages_df = pd.DataFrame(messages)
```

```python
messages_df.head(10)
```

```python
print("Shape:", messages_df.shape)

print(
    "Duplicate message IDs:",
    messages_df["message_id"].duplicated().sum()
)

print(
    "All platform IDs valid:",
    messages_df["platform_user_id"].isin(
        customers_df["platform_user_id"]
    ).all()
)

print(
    "All product IDs valid:",
    messages_df["current_product_id"].dropna().isin(
        products_df["product_id"]
    ).all()
)

print(messages_df["intent"].value_counts())
```

```python
messages_check = messages_df.merge(
    customers_df[
        ["platform_user_id", "created_at"]
    ],
    on="platform_user_id"
)

print(
    "Messages before customer creation:",
    (
        messages_check["timestamp"]
        <
        messages_check["created_at"]
    ).sum()
)
```

```python
customers_df.to_csv("customers.csv", index=False)
products_df.to_csv("products.csv", index=False)
variants_df.to_csv("product_variants.csv", index=False)
orders_df.to_csv("orders.csv", index=False)
order_items_df.to_csv("order_items.csv", index=False)
messages_df.to_csv("messages.csv", index=False)
```
